### Cleaning Data (Pembersihan Data)
ini merupakan tahapan penghapusan atau koreksi data yang bermasalah untuk memastikan data siap dianalisa

```python
# 1. Konversi Tipe Data
# Mengubah kolom 'dteday' menjadi tipe datetime
df_hour['dteday'] = pd.to_datetime(df_hour['dteday'])
df_day['dteday'] = pd.to_datetime(df_day['dteday'])
```

```python
# 2. Mapping Kolom Kategori ke Label yang Jelas
season_map = {1: 'Spring', 2: 'Summer', 3: 'Fall', 4: 'Winter'}
weathersit_map = {
    1: 'Cerah/Sedikit berawan/Sebagian berawan',
    2: 'Berkabut + berawan/kabut + awan terputus/kabut + sedikit awan/Berkabut',
    3: 'Hujan ringan/hujan ringan + badai petir + awan tersebar/hujan ringan + awan',
    4: 'Hujan lebat + hujan es + badai petir + kabut/salju + kabut'
}
weekday_map = {0: 'Minggu', 1: 'Senin', 2: 'Selasa', 3: 'Rabu', 4: 'Kamis', 5: 'Jumat', 6: 'Sabtu'}

# Terapkan mapping pada df_day (umumnya digunakan untuk analisis harian)
df_day['year_label'] = df_day['yr'].map({0: 2011, 1: 2012})
df_day['season_label'] = df_day['season'].map(season_map)
df_day['weathersit_label'] = df_day['weathersit'].map(weathersit_map)
df_day['weekday_label'] = df_day['weekday'].map(weekday_map)

# Terapkan year_label pada df_hour
df_hour['year_label'] = df_hour['yr'].map({0: 2011, 1: 2012})
```

```python
# 3. Menghapus Kolom Redundan/Tidak Perlu (instant, yr)
# yr sudah diwakili oleh year_label
df_day = df_day.drop(columns=["instant", "yr"])
df_hour = df_hour.drop(columns=["instant", "yr"])
```

```python
# Cek hasil cleaning
print("Informasi df_day setelah cleaning:")
print(df_day.info())
df_day
```

## Visualization & Explanatory Analysis

## Pertanyaan 1:
Bagaimana pola rata-rata permintaan penyewaan sepeda sepanjang hari?

```python
jam = pd.read_csv("hour.csv")
Rata_per_jam = jam.groupby("hr")["cnt"].mean().reset_index()
Rata_per_jam
```

```python
plt.plot(
    Rata_per_jam["hr"],
    Rata_per_jam["cnt"],
    marker ='x'
)
plt.xlabel("Jam")
plt.ylabel("Jumlah Penyewaan")
plt.title("Rata-rata Penyewaan Sepeda per Jam")
plt.show()
```

## Pertanyaan 2:
Bagaimana rata-rata jumlah penyewaan sepeda pada setiap musim?


```python
day = pd.read_csv("day.csv")
Rata_per_season = day.groupby("season")["cnt"].mean().reset_index()
Rata_per_season
```

```python
# Mapping angka musim ke nama musim
season_names = {
    1: "Spring",
    2: "Summer",
    3: "Fall",
    4: "Winter"
}

# Tambahkan kolom baru dengan nama musim
Rata_per_season["season_name"] = Rata_per_season["season"].map(season_names)

# Plot grafik rata-rata penyewaan per musim
plt.figure(figsize=(6,4))
plt.bar(
    Rata_per_season["season_name"],
    Rata_per_season["cnt"]
)

plt.title("Rata-rata Penyewaan Sepeda per Musim")
plt.xlabel("Musim")
plt.ylabel("Jumlah Penyewaan")
plt.grid(axis="y")
plt.show()
```

## Pertanyaan 3:
Berapa rata-rata tertinggi total penyewaan sepeda pada setiap kategori hari, dalam setiap tahun?

```python
def categorize(row):
    if row["holiday"] == 1:
        return "Holiday"
    elif row["workingday"] == 1:
        return "Working"
    else:
        return "Weekday"

day["day_category"] = day.apply(categorize, axis=1)

# Konversi tahun
day["year"] = day["yr"].map({0: 2011, 1: 2012})

# Hitung rata-rata per kategori hari dan tahun
Rata_per_year_cat = day.groupby(["year", "day_category"])["cnt"].mean().reset_index()

# Pivot tabel 1 tampilannya
Rata_table = Rata_per_year_cat.pivot(index="year", columns="day_category", values="cnt")

Rata_table
```

```python
# Grafik Tahun 2011
plt.figure(figsize=(7,5))
subset_2011 = Rata_per_year_cat[Rata_per_year_cat["year"] == 2011]

plt.bar(
    subset_2011["day_category"],
    subset_2011["cnt"],
    alpha=0.7
)

plt.title("Rata-rata Penyewaan Sepeda per Kategori Hari (2011)")
plt.xlabel("Kategori Hari")
plt.ylabel("Jumlah Penyewaan")
plt.grid(axis="y")
plt.show()


# Grafik Tahun 2012
plt.figure(figsize=(7,5))
subset_2012 = Rata_per_year_cat[Rata_per_year_cat["year"] == 2012]

plt.bar(
    subset_2012["day_category"],
    subset_2012["cnt"],
    alpha=0.7
)

plt.title("Rata-rata Penyewaan Sepeda per Kategori Hari (2012)")
plt.xlabel("Kategori Hari")
plt.ylabel("Jumlah Penyewaan")
plt.grid(axis="y")
plt.show()
```

## Pertanyaan 4:
Berapa Total Sepeda yang disewakan pada kategori hari (workingday) disetiap musimnya?

```python
total_permusim = day[day['workingday'] == 1].groupby('season')['cnt'].sum()
total_permusimdf = total_permusim.reset_index()
total_permusimdf
```

```python
# Plot bar
labels = ['Springer', 'Summer', 'Fall', 'Winter']
plt.bar(labels, total_permusim.values)
plt.xlabel('Musim')
plt.ylabel('Total Penyewaan')
plt.title('Total Penyewaan Sepeda pada Working Day per Musim')
plt.show()
```

## Pertanyaan 5:
Total penyewaan sepeda setiap tahunnya, dan buatkan visualisasinya menggunakan grafik?

```python
total_per_tahun = day.groupby('yr')['cnt'].sum()
total_per_tahundf = total_per_tahun.reset_index()
total_per_tahundf
```

```python
#Visualisasi menggunakan bar plot
labels = ['2011', '2012']
plt.bar(labels, total_per_tahun.values)
plt.xlabel('Tahun')
plt.ylabel('Total Penyewaan')
plt.title('Total Penyewaan Sepeda pada tiap Tahun')
plt.show()
```

## Pertanyaan 6:
Manakah tipe pengguna yang lebih banyak menyewa sepeda secara keseluruhan (casual atau registed)?

```python
total_casual = day['casual'].sum()
total_registered = day['registered'].sum()
tipe_user = ['Casual', 'Registered']
totals = [total_casual, total_registered]
print(f"Total Casual: {total_casual}")
print(f"Total Registered: {total_registered}")
```

```python
plt.figure(figsize=(6, 6))
plt.pie(totals, labels=tipe_user, autopct='%1.1f%%', colors=['skyblue', 'orange'], startangle=90)
plt.title('Persentase Penyewaan: Casual vs Registered')
plt.show()
```
